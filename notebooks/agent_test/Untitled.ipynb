```python
from langchain_neo4j import Neo4jGraph
```

```python

# Neo4j 연결 설정
uri = "bolt://neo4j-gds-apoc-n10s:7687"  # Neo4j의 Bolt 프로토콜 URI
username = "neo4j"  # 사용자 이름
password = "neo4jpassword"  # 비밀번호

# Neo4j 드라이버 생성
graph = Neo4jGraph(
            url = "bolt://neo4j-gds-apoc-n10s:7687",
            username="neo4j",
            password="neo4jpassword",
            enhanced_schema=True,
            sanitize=True
        )
```

```python
target_plan = "0 청년 59"
```

```python
query = f"""
MATCH (p:`요금제`)
WHERE p.`상품명` = '{target_plan}'  // p를 특정하려면 파라미터 또는 조건이 필요함
OPTIONAL MATCH (p)-[r]-(n)
WITH p, collect({{
    relationship: {{
        type: type(r),
        properties: properties(r)
    }},
    node: {{
        labels: labels(n),
        properties: properties(n)
    }}
}}) AS neighbors
RETURN [{{
    p: {{
        labels: labels(p),
        properties: properties(p)
    }},
    neighbors: neighbors
}}] AS result
"""
```

```python
query_res = graph.query(query)
```
