```python
customer_info_list = []
for prod in prod_info:
    pmProductId = prod["pmProductID"]
    
    customerInfo = prod.get("customerInfo", {})
    
    onboardingCustomerWelfareType = customerInfo.get("onboardingCustomerWelfareType", {})
    welfarediscounttypeavailability = onboardingCustomerWelfareType.get("welfareDiscountTypeAvailability", {})
    
    onboardingCustomer = customerInfo.get("onboardingCustomer", {})
    
    agerule_raw = onboardingCustomer.get("ageRule", [])
    agerule_from = 0
    agerule_to = 999
    for _rule in agerule_raw:
        if _rule.get("eligibility", {}):
            _age = regex.findall(_rule["value"])
            if _age:
                if "이상" in _rule["value"]:
                    agerule_from = int(_age[0])
                elif "이하" in _rule["value"]:
                    agerule_to = int(_age[0])
        else:
            print("??")
    agerule = [agerule_from, agerule_to]
    
    businesscustomersubtyperule_raw = onboardingCustomer.get("businessCustomerSubtypeRule", {})
    if businesscustomersubtyperule_raw.get("eligibility", {}):
        businesscustomersubtyperule = businesscustomersubtyperule_raw.get("valueList", [])
    
    customertyperule_raw = onboardingCustomer.get("customerTypeRule", {})
    if customertyperule_raw.get("eligibility", {}):
        customertyperule = customertyperule_raw.get("valueList", [])
    
    individualcustomersubtyperule_raw = onboardingCustomer.get("individualCustomerSubtypeRule", {})
    if individualcustomersubtyperule_raw.get("eligibility", {}):
        individualcustomersubtyperule = individualcustomersubtyperule_raw.get("valueList")
    
    organizationcustomersubtyperule = onboardingCustomer.get("organizationCustomerSubtypeRule", {})
    
    customer_info_list.append([
        pmProductId, welfarediscounttypeavailability, agerule, businesscustomersubtyperule, customertyperule, individualcustomersubtyperule, organizationcustomersubtyperule
    ])
```

```python
for x in customer_info_list:
    print(x[0], x[2])
```
